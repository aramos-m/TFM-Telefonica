```python
import os
from moviepy.editor import VideoFileClip

# Carpeta actual donde está el notebook
carpeta_actual = os.getcwd()

# Carpeta gemela "videos"
carpeta_videos = os.path.abspath(os.path.join(carpeta_actual, "..", "data"))
# Carpeta de salida "audio" dentro de la carpeta del notebook
carpeta_audio = os.path.join(carpeta_actual, "audio")
os.makedirs(carpeta_audio, exist_ok=True)

# Buscar el primer mp4 en la carpeta videos
mp4_files = [f for f in os.listdir(carpeta_videos) if f.lower().endswith(".mp4")]
if not mp4_files:
    raise FileNotFoundError("No se encontró ningún archivo .mp4 en la carpeta 'videos'.")

archivo_mp4 = mp4_files[0]  # tomar el primero
ruta_video = os.path.join(carpeta_videos, archivo_mp4)

# Crear ruta de salida con mismo nombre pero extensión .wav
nombre_salida = os.path.splitext(archivo_mp4)[0] + ".wav"
ruta_salida = os.path.join(carpeta_audio, nombre_salida)

# Conversión
print(f"→ Convirtiendo: {ruta_video}")
video = VideoFileClip(ruta_video)
audio = video.audio
audio.write_audiofile(ruta_salida)
video.close()
audio.close()
print(f"✔ Guardado en: {ruta_salida}")
```
